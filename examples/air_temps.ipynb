```python
import pandas as pd
import matplotlib.pyplot as plt
from geoemerge import globe_data
```

```python
print("Fetching air temperature data from GLOBE...")

gdf = globe_data(
    protocol="air_temps",
    start_date="2025-05-01",
    end_date="2025-05-02",
    country_code="DOM"
)
```

```python
# Display the cleaned data
print("\n--- Data Overview ---")
display_cols = ['MeasuredAt', 'SiteName', 'CurrentTemp', 'Elevation']
print(gdf[display_cols].head(10))

# Plot the temperature curve for a specific site
site_name = "GRI GRI COURT:ATM-06"
site_data = gdf[gdf['SiteName'] == site_name]

if not site_data.empty:
    plt.figure(figsize=(10, 5))
    plt.plot(site_data['MeasuredAt'], site_data['CurrentTemp'], marker='o', linestyle='-', color='b')
    plt.title(f'Air Temperature Over Time\n{site_name}')
    plt.xlabel('Time')
    plt.ylabel('Temperature (°C)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
